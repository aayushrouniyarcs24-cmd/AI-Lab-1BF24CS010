```python
import pandas as pd
import itertools

def evaluate(expr, model):
    return eval(expr, {}, model)

def generate_interactive_truth_table(symbols, kb, alpha):
    rows = []
    num_symbols = len(symbols)

    for truth_values in itertools.product([True, False], repeat=num_symbols):
        model = dict(zip(symbols, truth_values))

        kb_evals = [evaluate(clause, model) for clause in kb]
        kb_true = all(kb_evals)

        alpha_true = evaluate(alpha, model)
        entails = (not kb_true) or alpha_true

        row = {sym: model[sym] for sym in symbols}
        row.update({
            'KB': kb_true,
            'alpha (Query)': alpha_true,
            'KB => alpha': entails
        })
        rows.append(row)

    df = pd.DataFrame(rows)

    kb_true_rows = df[df['KB'] == True]
    if kb_true_rows.empty:
        overall_entails = True
    else:
        overall_entails = kb_true_rows['alpha (Query)'].all()

    return df, overall_entails

if __name__ == "__main__":
    print("=== Propositional Logic Entailment Verification Tool ===")
    print("Note: Use standard Python syntax operators: 'and', 'or', 'not'")
    print("Example implication (p -> q) must be entered as: 'not p or q'\n")

    raw_symbols = input("Enter all proposition symbols separated by commas (e.g., p, q, r): ")
    symbols = [sym.strip() for sym in raw_symbols.split(",") if sym.strip()]

    print("\nEnter your Knowledge Base (KB) clauses one by one.")
    print("Type 'done' when you are finished entering rules.")
    kb_clauses = []
    while True:
        clause = input(f"Enter KB Clause #{len(kb_clauses) + 1}: ").strip()
        if clause.lower() == 'done':
            if not kb_clauses:
                print("Error: The KB cannot be completely empty. Please add a rule.")
                continue
            break
        if clause:
            kb_clauses.append(clause)

    query = input("\nEnter the target Query expression (alpha): ").strip()

    print("\n[Processing Execution Matrix...]")
    df_result, does_entail = generate_interactive_truth_table(symbols, kb_clauses, query)

    print("\n=== GENERATED TRUTH TABLE ===")
    print(df_result.to_string(index=False))
    print("==============================")

    print(f"\nFinal Result: Does KB entail '{query}'? -> **{'YES' if does_entail else 'NO'}**\n")
```
