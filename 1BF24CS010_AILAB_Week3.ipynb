```python
from collections import deque

# =====================================================
# INITIAL AND GOAL STATE
# =====================================================

INITIAL = (
    2, 8, 3,
    1, 6, 4,
    7, 0, 5
)

GOAL = (
    1, 2, 3,
    8, 0, 4,
    7, 6, 5
)


# =====================================================
# PRINT STATE
# =====================================================

def print_state(state):
    print(" ", state[0], state[1], state[2])
    print(" ", state[3], state[4], state[5])
    print(" ", state[6], state[7], state[8])


# =====================================================
# MANHATTAN DISTANCE
# =====================================================

def manhattan_distance(state):
    distance = 0
    for i in range(9):
        tile = state[i]

        # Ignore blank
        if tile == 0:
            continue

        # Current position
        current_row = i // 3
        current_col = i % 3

        # Goal position
        goal_index = GOAL.index(tile)
        goal_row = goal_index // 3
        goal_col = goal_index % 3

        # Manhattan distance
        distance += abs(current_row - goal_row)
        distance += abs(current_col - goal_col)

    return distance


# =====================================================
# GENERATE CHILDREN
# =====================================================

def generate_children(state):
    children = []
    blank = state.index(0)
    row = blank // 3
    col = blank % 3

    moves = [
        ("UP", -1, 0),
        ("DOWN", 1, 0),
        ("LEFT", 0, -1),
        ("RIGHT", 0, 1)
    ]

    for move, dr, dc in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_blank = new_row * 3 + new_col
            new_state = list(state)
            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]
            children.append(
                (move, tuple(new_state))
            )

    return children


# =====================================================
# PRINT NODE
# =====================================================

def print_node(state, g, move):
    h = manhattan_distance(state)
    f = g + h

    print(" Move:", move)
    print_state(state)
    print(" g(n) =", g)
    print(" h(n) =", h)
    print(" f(n) =", f)

    if state == GOAL:
        print(" >>> GOAL <<<")
    print()


# =====================================================
# PRINT STATE SPACE
# =====================================================

def print_state_space():
    print("\n")
    print("=" * 60)
    print("A* STATE SPACE - MANHATTAN DISTANCE")
    print("=" * 60)

    print("\nINITIAL STATE")
    print_state(INITIAL)

    h = manhattan_distance(INITIAL)
    print("\ng(n) =", 0)
    print("h(n) =", h)
    print("f(n) =", h)
    print("\n" + "=" * 60)

    queue = deque()
    queue.append(
        (INITIAL, 0, [])
    )

    visited = {INITIAL}
    goal_found = False

    while queue:
        current, depth, path = queue.popleft()

        if current == GOAL:
            continue

        print("\n")
        print("-" * 60)
        print("DEPTH =", depth)
        print("-" * 60)

        print("\nCURRENT STATE:")
        print_state(current)

        h = manhattan_distance(current)
        print(
            "\ng =", depth,
            " h =", h,
            " f =", depth + h
        )

        print("\nPOSSIBLE MOVES:")
        children = generate_children(current)

        for move, child in children:
            if child in visited:
                continue

            visited.add(child)
            print("\n-----------------------------")
            print_node(
                child,
                depth + 1,
                move
            )

            queue.append(
                (
                    child,
                    depth + 1,
                    path + [move]
                )
            )

            if child == GOAL:
                goal_found = True
                print("Solution path:")
                print(
                    " -> ".join(
                        path + [move]
                    )
                )
                print("\nGOAL REACHED!")

        if goal_found:
            break


# =====================================================
# MAIN
# =====================================================

print_state_space()
```

```python


from collections import deque

# =====================================================
# INITIAL AND GOAL STATE
# =====================================================

INITIAL = (
    2, 8, 3,
    1, 6, 4,
    7, 0, 5
)

GOAL = (
    1, 2, 3,
    8, 0, 4,
    7, 6, 5
)


# =====================================================
# PRINT A STATE
# =====================================================

def print_state(state):
    print(" ", state[0], state[1], state[2])
    print(" ", state[3], state[4], state[5])
    print(" ", state[6], state[7], state[8])


# =====================================================
# MISPLACED TILES
# =====================================================

def misplaced_tiles(state):
    h = 0
    for i in range(9):
        # Do not count blank
        if state[i] != 0 and state[i] != GOAL[i]:
            h += 1
    return h


# =====================================================
# GENERATE ALL POSSIBLE MOVES
# =====================================================

def generate_children(state):
    children = []
    blank = state.index(0)
    row = blank // 3
    col = blank % 3

    moves = [
        ("UP", -1, 0),
        ("DOWN", 1, 0),
        ("LEFT", 0, -1),
        ("RIGHT", 0, 1)
    ]

    for move, dr, dc in moves:
        new_row = row + dr
        new_col = col + dc

        # Check whether move is possible
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_blank = new_row * 3 + new_col
            new_state = list(state)
            # Swap blank
            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]
            children.append(
                (move, tuple(new_state))
            )

    return children


# =====================================================
# PRINT NODE INFORMATION
# =====================================================

def print_node(state, g, move):
    h = misplaced_tiles(state)
    f = g + h

    print(" Move:", move)
    print_state(state)
    print(" g(n) =", g)
    print(" h(n) =", h)
    print(" f(n) =", f)

    if state == GOAL:
        print(" >>> GOAL <<<")
    print()


# =====================================================
# PRINT COMPLETE STATE SPACE
# =====================================================

def print_state_space():
    print("\n")
    print("=" * 60)
    print("A* STATE SPACE - MISPLACED TILES")
    print("=" * 60)

    print("\nINITIAL STATE")
    print_state(INITIAL)

    print("\ng(n) =", 0)
    print("h(n) =", misplaced_tiles(INITIAL))
    print("f(n) =", misplaced_tiles(INITIAL))
    print("\n" + "=" * 60)

    # Queue stores:
    # state, depth, path
    queue = deque()
    queue.append(
        (INITIAL, 0, [])
    )

    # Visited states
    visited = {INITIAL}
    goal_found = False

    while queue:
        current, depth, path = queue.popleft()

        # Stop after generating goal
        if current == GOAL:
            continue

        print("\n")
        print("-" * 60)
        print("DEPTH =", depth)
        print("-" * 60)

        print("\nCURRENT STATE:")
        print_state(current)
        print(
            "\ng =", depth,
            " h =", misplaced_tiles(current),
            " f =", depth + misplaced_tiles(current)
        )

        print("\nPOSSIBLE MOVES:")
        children = generate_children(current)

        for move, child in children:
            # Don't revisit states
            if child in visited:
                continue

            visited.add(child)
            print("\n-----------------------------")
            print_node(
                child,
                depth + 1,
                move
            )

            # Add child to queue
            queue.append(
                (
                    child,
                    depth + 1,
                    path + [move]
                )
            )

            # If goal reached
            if child == GOAL:
                goal_found = True
                print("Solution path:")
                print(
                    " -> ".join(
                        path + [move]
                    )
                )
                print("\nGOAL REACHED!")

        if goal_found:
            break


# =====================================================
# MAIN
# =====================================================

print_state_space()
```
