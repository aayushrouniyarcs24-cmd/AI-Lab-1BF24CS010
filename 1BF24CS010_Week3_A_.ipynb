```python
import heapq


INITIAL = (
    2, 8, 3,
    1, 6, 4,
    7, 0, 5
)

GOAL = (
    1, 2, 3,
    8, 0, 4,
    7, 6, 5
)


def print_puzzle(state):

    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])

    print()



def manhattan_distance(state):

    distance = 0

    for i in range(9):

        tile = state[i]

        # Don't calculate distance for blank
        if tile == 0:
            continue

        # Current position
        current_row = i // 3
        current_col = i % 3

        # Goal position
        goal_index = GOAL.index(tile)

        goal_row = goal_index // 3
        goal_col = goal_index % 3

        # Manhattan distance
        distance += abs(current_row - goal_row)
        distance += abs(current_col - goal_col)

    return distance



def get_neighbors(state):

    neighbors = []

    # Find blank
    blank = state.index(0)

    row = blank // 3
    col = blank % 3

    moves = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1)     # Right
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        # Check valid position
        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank = new_row * 3 + new_col

            new_state = list(state)

            # Swap blank and tile
            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            neighbors.append(tuple(new_state))

    return neighbors

def a_star():

    # Priority queue
    # (f, g, state)
    open_list = []

    h = manhattan_distance(INITIAL)

    heapq.heappush(
        open_list,
        (h, 0, INITIAL)
    )

    # Cost from initial state
    g_cost = {
        INITIAL: 0
    }

    # Parent dictionary
    parent = {
        INITIAL: None
    }

    while open_list:

        f, g, current = heapq.heappop(open_list)

        # Ignore outdated entries
        if g != g_cost[current]:
            continue

        # Goal reached
        if current == GOAL:

            # Reconstruct path
            path = []

            while current is not None:
                path.append(current)
                current = parent[current]

            path.reverse()

            print("\n===== SOLUTION =====")
            print("Total moves:", len(path) - 1)

            for step, state in enumerate(path):

                g_value = step
                h_value = manhattan_distance(state)
                f_value = g_value + h_value

                print("\nStep:", step)
                print("g(n) =", g_value)
                print("h(n) =", h_value)
                print("f(n) =", f_value)

                print_puzzle(state)

            return

        # Generate successors
        for neighbor in get_neighbors(current):

            new_g = g + 1

            # Check whether better path is found
            if neighbor not in g_cost or new_g < g_cost[neighbor]:

                g_cost[neighbor] = new_g
                parent[neighbor] = current

                h = manhattan_distance(neighbor)
                f = new_g + h

                heapq.heappush(
                    open_list,
                    (f, new_g, neighbor)
                )

    print("No solution found.")



print("===== A* SEARCH =====")
print("Heuristic: Manhattan Distance")

print("\nInitial State:")
print_puzzle(INITIAL)

print("Goal State:")
print_puzzle(GOAL)

a_star()


import heapq

# Goal state
GOAL = (
    1, 2, 3,
    8, 0, 4,
    7, 6, 5
)

# Initial state
INITIAL = (
    2, 8, 3,
    1, 6, 4,
    7, 0, 5
)



def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])
    print()

def misplaced_tiles(state):
    count = 0

    for i in range(9):

        # Don't count blank
        if state[i] != 0 and state[i] != GOAL[i]:
            count += 1

    return count



def get_neighbors(state):

    neighbors = []

    # Find blank
    blank = state.index(0)

    row = blank // 3
    col = blank % 3

    # Possible moves
    moves = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1)     # Right
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        # Check valid position
        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank = new_row * 3 + new_col

            # Convert tuple to list
            new_state = list(state)

            # Swap blank and tile
            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            neighbors.append(tuple(new_state))

    return neighbors


def a_star():

    # Priority queue
    # (f, g, state)
    open_list = []

    h = misplaced_tiles(INITIAL)

    heapq.heappush(
        open_list,
        (h, 0, INITIAL)
    )

    # Store best g value
    g_cost = {
        INITIAL: 0
    }

    # Store parent of each state
    parent = {
        INITIAL: None
    }

    while open_list:

        f, g, current = heapq.heappop(open_list)

        # Ignore old entries
        if g != g_cost[current]:
            continue

        # Goal test
        if current == GOAL:

            # Reconstruct path
            path = []

            while current is not None:
                path.append(current)
                current = parent[current]

            path.reverse()

            print("\n===== SOLUTION =====")
            print("Total moves:", len(path) - 1)

            for step, state in enumerate(path):

                g_value = step
                h_value = misplaced_tiles(state)
                f_value = g_value + h_value

                print("\nStep:", step)
                print("g(n) =", g_value)
                print("h(n) =", h_value)
                print("f(n) =", f_value)

                print_puzzle(state)

            return

        # Generate children
        for neighbor in get_neighbors(current):

            new_g = g + 1

            # If this is a better path
            if neighbor not in g_cost or new_g < g_cost[neighbor]:

                g_cost[neighbor] = new_g
                parent[neighbor] = current

                h = misplaced_tiles(neighbor)
                f = new_g + h

                heapq.heappush(
                    open_list,
                    (f, new_g, neighbor)
                )

    print("No solution found.")


# -----------------------------------------
# Main
# -----------------------------------------
print("===== A* SEARCH =====")
print("Heuristic: Misplaced Tiles")

print("\nInitial State:")
print_puzzle(INITIAL)

print("Goal State:")
print_puzzle(GOAL)

a_star()
```
